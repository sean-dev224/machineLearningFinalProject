***This file contains code to generate visualizations of data***

```python
import numpy as np
import pandas as pd
import seaborn
import matplotlib.pyplot as plt

DATA_PATH = "data/diabetes_binary_5050split_health_indicators_BRFSS2015.csv"
```

```python
health_data = pd.read_csv(DATA_PATH)
health_data
```

```python
#pairplot generated with a few random samples to speed up processing
seaborn.pairplot(health_data.sample(1000, random_state=42), hue="Diabetes_binary")
```

```python
corr_matrix = health_data.corr()

plt.figure(figsize=(20, 15))
seaborn.heatmap(corr_matrix, annot=True, cmap="flare")
plt.show()
```
